# Generating Tables

```python
import pandas as pd

def merge_data(fp_df, fn_df, mcc_df):
    fp = fp_df[["Model","Attack","Epsilon","FP","ASR (FP)","F1 Score","F1 Score (Adv)","FP (Adv)"]].copy()
    fp.columns = ["Model","Attack","Epsilon","Benign FP","ASR (FP)","Benign F1 (FP)","Adv F1 (FP)","Adv FP"]

    fn = fn_df[["Model","Attack","Epsilon","FN","ASR (FN)","F1 Score","F1 Score (Adv)","FN (Adv)"]].copy()
    fn.columns = ["Model","Attack","Epsilon","Benign FN","ASR (FN)","Benign F1 (FN)","Adv F1 (FN)","Adv FN"]

    mcc = mcc_df[["Model","Attack","Epsilon","ASR","MCC","MCC (Adv)"]].copy()
    mcc.columns = ["Model","Attack","Epsilon","ASR","MCC","Adv MCC"]

    merged = pd.merge(fp, fn, on=["Model","Attack","Epsilon"], how="inner")
    merged = pd.merge(merged, mcc, on=["Model","Attack","Epsilon"], how="inner")

    # choose which F1 to show (FP side here)
    merged["Benign F1"] = merged["Benign F1 (FP)"]
    merged["Adv F1"] = merged["Adv F1 (FP)"]

    merged = merged[[
        "Model","Benign FP","Benign FN", "MCC","Benign F1",
        "Attack","Epsilon",
        "Adv FP","ASR (FP)","Adv FN","ASR (FN)","Adv F1","Adv MCC", "ASR"
    ]]
    return merged.sort_values(["Model","Attack","Epsilon"])
```

```python
# def sort_table(model):
#     order = ["DT", "ET", "RF", "XGBoost", "DNN"]
#     model["Model"] = pd.Categorical(model["Model"], categories=order, ordered=True)
#     model = model.sort_values(["Model","Attack","Epsilon"]).reset_index(drop=True)
#     return model

def sort_table(df):
    order = ["DT", "ET", "RF", "XGBoost", "DNN"]
    df = df.copy()
    df["Model"] = pd.Categorical(df["Model"], categories=order, ordered=True)

    attack_order = ["FGSM", "BIM", "PGD"]
    df["Attack"] = pd.Categorical(df["Attack"], categories=attack_order, ordered=True)

    df = df.sort_values(["Model", "Attack", "Epsilon"]).reset_index(drop=True)
    return df


def convert_all_percents(df, decimals=3):
    df = df.copy()
    for col in df.columns:
        # only touch columns that contain % strings
        if df[col].astype(str).str.contains("%").any():
            df[col] = (df[col].astype(str)
                       .str.replace("%", "", regex=False)
                       .astype(float)
                       .div(100)
                       .round(decimals))
    return df
```

```python
import pandas as pd

fa_fp = pd.read_csv("results/fa_fp.csv")
fa_fn = pd.read_csv("results/fa_fn.csv")
fa_mcc = pd.read_csv("results/fa.csv")


mecta_fp = pd.read_csv("results/mecta_fp.csv")
mecta_fn = pd.read_csv("results/mecta_fn.csv")
mecta_mcc = pd.read_csv("results/mecta.csv")

msa_fp = pd.read_csv("results/msa_fp.csv")
msa_fn = pd.read_csv("results/msa_fn.csv")
msa_mcc = pd.read_csv("results/msa.csv")

csa_fp = pd.read_csv("results/csa_fp.csv")
csa_fn = pd.read_csv("results/csa_fn.csv")
csa_mcc = pd.read_csv("results/csa.csv")

rloffa_fp = pd.read_csv("results/rloffa_fp.csv")
rloffa_fn = pd.read_csv("results/rloffa_fn.csv")
rloffa_mcc = pd.read_csv("results/rloffa.csv")

rlona_fp = pd.read_csv("results/rlona_fp.csv")
rlona_fn = pd.read_csv("results/rlona_fn.csv")
rlona_mcc = pd.read_csv("results/rlona.csv")

csa_fp =pd.read_csv("results/csa.csv")



tab_fa = merge_data(fa_fp, fa_fn, fa_mcc)
tab_mecta = merge_data(mecta_fp, mecta_fn, mecta_mcc)
tab_msa = merge_data(msa_fp, msa_fn, msa_mcc)
tab_rloffa = merge_data(rloffa_fp, rloffa_fn, rloffa_mcc)
tab_rlona = merge_data(rlona_fp, rlona_fn, rlona_mcc)

tab_fa = sort_table(tab_fa)
tab_mecta = sort_table(tab_mecta)
tab_msa = sort_table(tab_msa)
tab_rloffa = sort_table(tab_rloffa)
tab_rlona= sort_table(tab_rlona)
csa_fp = sort_table(csa_fp)

tab_fa = convert_all_percents(tab_fa)
tab_mecta = convert_all_percents(tab_mecta)
tab_msa = convert_all_percents(tab_msa)
tab_rloffa = convert_all_percents(tab_rloffa)
tab_rlona= convert_all_percents(tab_rlona)
csa_fp = convert_all_percents(csa_fp)



tab_fa.to_csv("results/tableIII.csv", index =False)
tab_mecta.to_csv("results/tableIV.csv", index =False)
tab_msa.to_csv("results/tableV.csv", index =False)
tab_rloffa.to_csv("results/tableVI.csv", index =False)
tab_rlona.to_csv("results/tableVII.csv", index =False)

tab_mecta
# csa_fp.to_csv("results/tableVIII.csv", index=False)

# csa_fp
```

```python
def df_to_latex(df, caption, label):
    # expects columns:
    # Model, Benign FP, Benign FN, MCC, Attack, Epsilon, Adv FP, ASR (FP), Adv FN, ASR (FN), Adv MCC
    lines = []
    lines.append(r"\begin{table}[!t]")
    lines.append(r"\centering")
    lines.append(rf"\caption{{{caption}}}")
    lines.append(rf"\label{{{label}}}")
    lines.append(r"\scriptsize")
    lines.append(r"\setlength{\tabcolsep}{2.5pt}")
    lines.append(r"\renewcommand{\arraystretch}{1}")
    lines.append(r"\begin{tabular}{|c|ccc|c|c|cc|cc|c|}")
    lines.append(r"\hline")
    lines.append(r"\multirow{2}{*}{Model} & \multicolumn{3}{c|}{Benign} & \multirow{2}{*}{Attack} & \multirow{2}{*}{$\epsilon$} & \multicolumn{5}{c|}{Adversarial}\\")
    lines.append(r"\cline{2-4}\cline{7-11}")
    lines.append(r" & FP & FN & MCC &  &  & FP & ASR & FN & ASR & MCC\\")
    lines.append(r"\hline")

    for model, g in df.groupby("Model", sort=False):
        g = g.reset_index(drop=True)
        n = len(g)
        for i, row in g.iterrows():
            model_cell = rf"\multirow{{{n}}}{{*}}{{{model}}}" if i == 0 else ""
            line = "{} & {} & {} & {} & {} & {} & {} & {} & {} & {} & {}\\\\".format(
                model_cell,
                row["Benign FP"], row["Benign FN"], row["MCC"],
                row["Attack"], row["Epsilon"],
                row["Adv FP"], str(row["ASR (FP)"]).replace("%", r"\%"),
                row["Adv FN"], str(row["ASR (FN)"]).replace("%", r"\%"),
                row["Adv MCC"],
            )
            lines.append(line)
        lines.append(r"\hline")

    lines.append(r"\end{tabular}")
    lines.append(r"\end{table}")
    return "\n".join(lines)

latex_code_fa = df_to_latex(tab_fa, "Comparative analysis -- FA.", "tab:fa")
latex_code_mecta = df_to_latex(tab_mecta, "Comparative analysis -- MECTA.", "tab:mect")
latex_code_msa = df_to_latex(tab_msa, "Comparative analysis -- MSA.", "tab:msa")
latex_code_rloffa = df_to_latex(tab_rloffa, "Comparative analysis -- RLOFFA.", "tab:msa")
latex_code_rlona = df_to_latex(tab_rlona, "Comparative analysis -- RLONA.", "tab:msa")


```
