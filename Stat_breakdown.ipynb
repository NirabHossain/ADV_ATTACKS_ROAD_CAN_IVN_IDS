```python
# Define a helper function to count benign and malicious
def count_flags(csv_files):
    df = pd.read_csv(file_path)
    flag_col = df.columns[-1]  # use last column
    if df[flag_col].dtype == object:
        flags = df[flag_col].astype(str).str.strip()
        benign = (flags == '0').sum()
        malicious = (flags == '1').sum()
    else:
        benign = (df[flag_col] == 0).sum()
        malicious = (df[flag_col] == 1).sum()

    total = benign + malicious
    
    return (benign, malicious, total)
```

```python
def flag_value_count(name, flag_h): 
    return [name, *np.sum(flag_h, axis=0)]
```

```python
#Table I
import pandas as pd
import numpy as np
import glob
import os
csv_folder = 'preprocessed'  # Update this
csv_files = sorted(glob.glob(os.path.join(csv_folder, "*.csv")))
flag_h = []
for file_path in csv_files:
    result = count_flags(file_path)
    flag_h.append(result)

fl_csa =   flag_value_count("CSA", flag_h[:3])
fl_fa =    flag_value_count("FA", flag_h[3:6])
fl_mecta = flag_value_count("MECTA", flag_h[6:7])
fl_msa =   flag_value_count("MSA", flag_h[7:10])
fl_rloffa =flag_value_count("RLOFFA",flag_h[10:13])
fl_rlona = flag_value_count("RLONA",flag_h[13:16])
fl_total = flag_value_count("Total",flag_h)
lst=[fl_fa,fl_mecta,fl_msa,fl_rloffa,fl_rlona,fl_csa,fl_total]

summary_data = {'Attack Type': [], 'Benign': [], 'Malicious': [],'Total': []}
for name, benign, malicious, total in lst:
    summary_data['Attack Type'].append(name)
    summary_data['Benign'].append(benign)
    summary_data['Malicious'].append(malicious)
    summary_data['Total'].append(total)
df = pd.DataFrame(summary_data)

print("TABLE I\n")
print(df)
df.to_csv("results/tableI.csv", index=False)
for col in ["Benign", "Malicious", "Total"]:
    df[col] = df[col].map(lambda x: f"{int(x):,}")
df.loc[df["Attack Type"] == "Total", ["Attack Type","Benign","Malicious","Total"]] = \
    df.loc[df["Attack Type"] == "Total", ["Attack Type","Benign","Malicious","Total"]].map(lambda x: f"\\textbf{{{x}}}")
```
